```python
mark=np.array([[85,78,90],
      [92,85,98],
      [76,82,80]])
print(np.where(mark==78))
print(mark)
```

```python
mark=np.array([[85,78,90],
      [92,85,98],
      [76,82,80]])
print(np.where(mark>=80))
print(mark)
```

```python
mark=np.array([[85,78,90],
      [92,85,98],
      [76,82,80]])
print(mark[np.where(mark>=78)])
print(mark)
```
